```python
# =====================================================
# IMPORTS
# =====================================================

import pandas as pd
import numpy as np
import gc
import torch
import torch.nn as nn

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix
)

from sklearn.utils.class_weight import compute_class_weight

from datasets import Dataset

from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    TrainingArguments,
    Trainer
)
```

```python
# =====================================================
# LOAD BALANCED DATASETS
# =====================================================

course_balanced = pd.read_csv(
    "course_balanced-new.csv"
)

teacher_balanced = pd.read_csv(
    "teacher_balanced-new.csv"
)

university_balanced = pd.read_csv(
    "university_balanced-new.csv"
)

# =====================================================
# LOAD SHARED FIXED TEST SETS
# =====================================================

teacher_test = pd.read_csv(
    "teacher_test.csv"
)

course_test = pd.read_csv(
    "course_test.csv"
)

university_test = pd.read_csv(
    "university_test.csv"
)
```

```python
train_df = pd.concat(
    [teacher_balanced, university_balanced]
).reset_index(drop=True)

test_df = course_test.reset_index(drop=True)

print("Train:", train_df.shape)
print("Test:", test_df.shape)
```

```python
# =====================================================
# TRAIN / VALIDATION SPLIT
# =====================================================

train_df, val_df = train_test_split(

    train_df,

    test_size=0.1111,

    stratify=train_df["sentiment"],

    random_state=42
)

train_df = train_df.reset_index(drop=True)
val_df = val_df.reset_index(drop=True)

# FIXED SHARED TEST SET
test_df = test_df.reset_index(drop=True)

print("Train:", train_df.shape)
print("Validation:", val_df.shape)
print("Test:", test_df.shape)
```

```python
# =====================================================
# LABEL MAPPING
# =====================================================

label2id = {
    "negative": 0,
    "neutral": 1,
    "positive": 2
}

id2label = {
    0: "negative",
    1: "neutral",
    2: "positive"
}

train_df["label"] = train_df["sentiment"].map(label2id)
val_df["label"] = val_df["sentiment"].map(label2id)
test_df["label"] = test_df["sentiment"].map(label2id)
```

```python
# =====================================================
# FORMAT PROMPTS
# =====================================================

def format_instruction(row):

    return f"""<s>[INST] You are an expert in aspect-based sentiment analysis.

Classify the sentiment into:
positive, negative, or neutral.

Review:
{row['review']}

Aspect:
{row['aspect']}

Sentiment:
[/INST] {row['sentiment']}"""

train_df["text"] = train_df.apply(format_instruction, axis=1)
val_df["text"] = val_df.apply(format_instruction, axis=1)
test_df["text"] = test_df.apply(format_instruction, axis=1)
```

```python
# =====================================================
# CLASS WEIGHTS
# =====================================================

train_labels = train_df["label"].values

classes = np.unique(train_labels)

class_weights = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=train_labels
)

full_weights = torch.ones(3)

for cls, w in zip(classes, class_weights):

    full_weights[int(cls)] = w

class_weights = full_weights

print(class_weights)
```

```python
# =====================================================
# TOKENIZER
# =====================================================

model_name = "meta-llama/Meta-Llama-3-8B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(model_name)

tokenizer.pad_token = tokenizer.eos_token
```

```python
# =====================================================
# TOKENIZATION
# =====================================================

def tokenize_function(example):

    full_text = example["text"]
    label = example['label']

    prompt, answer = full_text.split("[/INST]")
    prompt = prompt + "[/INST]"
    answer = answer.strip()

    prompt_ids = tokenizer(
        prompt,
        truncation=True,
        max_length=64
    )["input_ids"]

    answer_ids = tokenizer(
        answer,
        truncation=True,
        max_length=64
    )["input_ids"]

    input_ids = prompt_ids + answer_ids
    labels = [-100] * len(prompt_ids) + answer_ids

    max_len = 128
    input_ids = input_ids[:max_len]
    labels = labels[:max_len]

    padding_length = max_len - len(input_ids)

    input_ids += [tokenizer.pad_token_id] * padding_length
    labels += [-100] * padding_length

    attention_mask = [
        1 if token != tokenizer.pad_token_id else 0
        for token in input_ids
    ]

    return {
        "input_ids": input_ids,
        "attention_mask": attention_mask,
        "labels": labels,
        "class_label": example["label"]
    }
```

```python
# =====================================================
# DATASETS
# =====================================================

train_dataset = Dataset.from_pandas(
    train_df[["text", "label"]]
)

val_dataset = Dataset.from_pandas(
    val_df[["text", "label"]]
)

test_dataset = Dataset.from_pandas(
    test_df[["text", "label"]]
)

train_dataset = train_dataset.map(
    tokenize_function,
    remove_columns=["text"]
)

val_dataset = val_dataset.map(
    tokenize_function,
    remove_columns=["text"]
)

test_dataset = test_dataset.map(
    tokenize_function,
    remove_columns=["text"]
)

train_dataset.set_format(type="torch")
val_dataset.set_format(type="torch")
test_dataset.set_format(type="torch")
```

```python
# =====================================================
# METRICS
# =====================================================

def compute_metrics(eval_pred):

    logits, labels = eval_pred

    preds = np.argmax(logits, axis=1)

    return {

        "accuracy": accuracy_score(labels, preds),

        "f1_macro": f1_score(
            labels,
            preds,
            average="macro"
        ),

        "f1_weighted": f1_score(
            labels,
            preds,
            average="weighted"
        )
    }
```

```python
# =====================================================
# EXPERIMENT FUNCTION
# =====================================================

def run_experiment(

    experiment_name,
    learning_rate=1e-5,
    epochs=5,
    use_class_weights=True
):

    gc.collect()
    torch.cuda.empty_cache()

    # =====================================================
    # LOAD MODEL
    # =====================================================

    model = AutoModelForCausalLM.from_pretrained(

        model_name,
        dtype=torch.float16,
        device_map="auto",
        offload_buffers=True,
        low_cpu_mem_usage=True
    )
    model.config.attention_dropout = 0.1
    model.config.hidden_dropout = 0.1
    
    model.gradient_checkpointing_enable()
    model.config.use_cache = False
    model.enable_input_require_grads()

    # =====================================================
    # TRAINER
    # =====================================================

    class WeightedTrainer(Trainer):

        def compute_loss(
            self,
            model,
            inputs,
            return_outputs=False,
            **kwargs
        ):

            labels = inputs.pop("labels")
            class_label = inputs.pop("class_label")

            outputs = model(**inputs)
            logits = outputs.logits

            shift_logits = logits[:, :-1, :]
            shift_labels = labels[:, 1:]

            loss_fct = nn.CrossEntropyLoss(
                reduction="none"
            )

            loss = loss_fct(
                shift_logits.reshape(
                    -1,
                    shift_logits.size(-1)
                ),
                shift_labels.reshape(-1)
            )

            loss = loss.view(shift_labels.size())

            mask = shift_labels != -100
            loss = loss * mask

            if use_class_weights:

                weights = class_weights.to(
                    class_label.device
                )[class_label].unsqueeze(1)

                loss = loss * weights

            loss = loss.sum() / mask.sum()

            return (
                (loss, outputs)
                if return_outputs
                else loss
            )

    # =====================================================
    # TRAINING ARGUMENTS
    # =====================================================

    training_args = TrainingArguments(

        output_dir=f"./{experiment_name}",

        per_device_train_batch_size=1,

        gradient_accumulation_steps=32,

        num_train_epochs=epochs,

        learning_rate=learning_rate,

        logging_steps=20,

        save_strategy="no",

        fp16=False,
        gradient_checkpointing=True,
        

        report_to="none",

        remove_unused_columns=False,

        optim="paged_adamw_8bit",

    )
    model.gradient_checkpointing_enable()
    model.config.use_cache = False
    model.train()

    # =====================================================
    # TRAINER
    # =====================================================

    trainer = WeightedTrainer(

        model=model,

        args=training_args,

        train_dataset=train_dataset,

        eval_dataset=val_dataset
    )

    # =====================================================
    # TRAIN
    # =====================================================

    trainer.train()

    # =====================================================
    # PREDICTION FUNCTION
    # =====================================================

    def predict(df):

        model.eval()

        preds = []
        labels = []

        device = next(model.parameters()).device

        for i in range(len(df)):

            review = df.iloc[i]["review"]

            aspect = df.iloc[i]["aspect"]

            true_label = df.iloc[i]["sentiment"]

            prompt = f"""<s>[INST] You are an expert sentiment classifier.

    Classify the sentiment of the given aspect.
    
    Possible labels:
    positive
    negative
    neutral
    
    Rules:
    - positive = praise/satisfaction
    - negative = complaint/criticism
    - neutral = mixed/unclear/objective
    
    Only output ONE WORD.
    
    Review:
    {review}
    
    Aspect:
    {aspect}
    
    Answer:
    [/INST]"""

            inputs = tokenizer(
                prompt,
                return_tensors="pt"
            ).to(device)

            with torch.inference_mode():

                output = model.generate(

                    **inputs,

                    max_new_tokens=5,

                    do_sample=False,

                    temperature=0.0,

                    pad_token_id=tokenizer.eos_token_id
                )

            decoded = tokenizer.decode(
                output[0],
                skip_special_tokens=True
            )

            pred_text = decoded.split(
                "[/INST]"
            )[-1].strip().lower()

            if "positive" in pred_text:

                pred = "positive"

            elif "negative" in pred_text:

                pred = "negative"

            else:

                pred = "neutral"

            preds.append(pred)

            labels.append(true_label)

        return preds, labels

    # =====================================================
    # RUN PREDICTIONS
    # =====================================================

    preds_cross, labels_cross = predict(test_df)

    # =====================================================
    # RESULTS
    # =====================================================

    print("\n===== CROSS DOMAIN RESULTS =====")

    accuracy = accuracy_score(
        labels_cross,
        preds_cross
    )

    macro_f1 = f1_score(
        labels_cross,
        preds_cross,
        average="macro"
    )

    weighted_f1 = f1_score(
        labels_cross,
        preds_cross,
        average="weighted"
    )

    print("Accuracy:", accuracy)

    print("Macro F1:", macro_f1)

    print("Weighted F1:", weighted_f1)

    print("\nClassification Report:\n")

    print(
        classification_report(
            labels_cross,
            preds_cross
        )
    )

    # =====================================================
    # SAVE PREDICTIONS
    # =====================================================

    final_predictions_df = test_df.copy()

    final_predictions_df[
        "true_sentiment"
    ] = labels_cross

    final_predictions_df[
        "predicted_sentiment"
    ] = preds_cross

    final_predictions_df.to_csv(

        f"teacher_university_to_course_lr3e5/{experiment_name}_predictions.csv",

        index=False
    )

    # =====================================================
    # SAVE METRICS
    # =====================================================
    
    metrics_df = pd.DataFrame({
    
        "Metric": [
            "Accuracy",
            "Macro_F1",
            "Weighted_F1"
        ],
    
        "Score": [
    
            accuracy,
            macro_f1,
            weighted_f1
        ]
    })
    
    metrics_df.to_csv(
    
        f"teacher_university_to_course_lr3e5/{experiment_name}_metrics.csv",
    
        index=False
    )

    # =====================================================
    # SAVE MODEL
    # =====================================================
    
    model.save_pretrained(
    
        f"teacher_university_to_course_lr3e5/{experiment_name}_model"
    )
    
    tokenizer.save_pretrained(
    
        f"teacher_university_to_course_lr3e5/{experiment_name}_model"
    )

    # =====================================================
    # CLEANUP
    # =====================================================

    trainer.optimizer = None
    trainer.lr_scheduler = None

    del trainer
    del model

    gc.collect()

    torch.cuda.empty_cache()

    torch.cuda.ipc_collect()

    return {

        "accuracy": accuracy,

        "macro_f1": macro_f1,

        "weighted_f1": weighted_f1
    }
```
